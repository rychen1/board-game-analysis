# Corpus descriptive analysis v1

Reproducible view of `bga-analyze` outputs for the BGG Game JSONL corpus.

This is **descriptive / exploratory**. It is not causal, not a population
study, and not a ranking. `bgg_boardgames_v1` is a convenience / coverage
sample.

Specification: `docs/analysis-corpus-descriptive-v1.md`.


```python
from pathlib import Path

from IPython.display import Image, Markdown, display

from board_game_analysis.analysis.run import run_corpus_descriptive
from board_game_analysis.config import Settings
from board_game_analysis.ingestion.corpus.ids import corpus_spec


def repo_root() -> Path:
    here = Path.cwd().resolve()
    for candidate in (here, *here.parents):
        if (candidate / "pyproject.toml").is_file() and (
            candidate / "src" / "board_game_analysis"
        ).is_dir():
            return candidate
    raise FileNotFoundError("could not locate repository root")


ROOT = repo_root()
CORPUS_ID = "bgg_boardgames_v1"
SPEC = corpus_spec(CORPUS_ID)
JSONL = ROOT / "data" / "processed" / "boardgamegeek" / SPEC.jsonl_name
if not JSONL.is_file():
    JSONL = ROOT / "data" / "processed" / "boardgamegeek" / "corpus_v0.jsonl"
    CORPUS_ID = "bgg_boardgames_v0"
    print("v1 JSONL missing; using local v0 file if present.")

settings = Settings.model_validate({"data_dir": ROOT / "data"})
result = run_corpus_descriptive(
    settings,
    corpus_id=CORPUS_ID,
    jsonl_path=JSONL,
)
print(f"corpus:  {CORPUS_ID}")
print(f"jsonl:   {JSONL}")
print(f"report:  {result.report_path}")
print(f"corpus_payload_id: {result.corpus_payload_id}")
print(f"report_payload_id: {result.report_payload_id}")
```
